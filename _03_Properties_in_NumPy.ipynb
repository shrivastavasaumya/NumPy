```python
# Properties

arr = np.array([1, 2, 3, 4, 5])
print(arr.shape) # (dimensions : m x n)

arr2D = np.array([[1, 2, 9], [3, 4, 8], [5, 6, 7], [10, 11, 12]])

print(arr2D.shape)
print(arr2D.size)
print(arr2D.ndim)
print(arr2D.dtype)    # to create from scratch 



# returns a copy in form of new array. It does not modify the original array in-place.
float_arr = arr.astype(np.float64)  # to change a existing array
print(float_arr, float_arr.dtype) 
```
