# Operations on Arrays
## 1) Reshaping

```python
arr = np.array([[1, 2, 9], [3, 4, 8], [5, 6, 7], [10, 11, 12]])
print(arr, arr.shape)



reshaped = arr.reshape((6,2))
print(reshaped, reshaped.shape)


# NOTE : "Reshaping" can only be done when we have compatible numbers
#         Example: here we have total 12 numbers {i.e  in 3 x 4 arrays = 12} so, reshaping should also be in multiple of 12
#                                                                                             {6 x 2} or {3 x 4} or {12 x 1}
```

## 2) flatten

```python

flattened = arr.flatten()   # --> from 4D to 1D
print(flattened, flattened.shape)
```

## 5) Boolean indexing

```python
# Based on a particular condition we can filter out our output

arr_boolean = np.array([4, 2, 1, 8, 5])
print(arr_boolean[arr_boolean > 3])      # those with "> 3" got printed
print(arr_boolean[arr_boolean % 2 == 0])  # only evens
```

## 6) Slicing

```python
arr_slicing = np.array([4, 2, 1, 8, 5])
                         # similar to python slicing : [start : end : step]
print(arr_slicing[1:4])            
print(arr_slicing[1:])
print(arr_slicing[:4])
print(arr_slicing[::2])  # for alternate elements
```

```python

nums = [1, 2, 3, 4, 5]
sub_list = nums[1:3]
print(sub_list)
sub_list[0] = 200

print(sub_list)
print(nums)     # will remain same



arr_nums = np.array([1, 2, 3, 4, 5])
sub_arr = arr_nums[1:3]
print(sub_arr)

sub_arr[0] = 200

print(sub_arr)
print(arr_nums)  # both will reflect changes
```

```python
arr_nums = np.array([1, 2, 3, 4, 5])


# BUT TO AVOID CHANGES IN ORIGINAL ARRAY USE ".copy"
sub_arr = arr_nums[1:3].copy()        #---> we are copying the view of the sliced array into this variable, not the entire array
print(sub_arr)

sub_arr[0] = 200

print(sub_arr)
print(arr_nums)
```
