# Creating arrays from lists

```python
import numpy as np
```

```python
#create - from lists
arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr), arr.dtype)

print(arr, type(arr), arr.shape) # shape- "Dimentions" , here we have aa 1D list/array which will revert us the shape as "length" of the list


arr2 = np.array([1, 2, 3, 4, 5, "cool", "prime"])
print(arr2, type(arr2), arr2.dtype)
print(arr2, type(arr), arr2.shape)
```

```python
# creating 2D array from 2D list
arr2D = np.array([[1, 2, 9], [3, 4, 8], [5, 6, 7], [10, 11, 12]])
print(arr2D, arr2D.shape)

#Shape = "Dimention" (m*n) value i.e rows * column value 
#Imagine a grid with these values (4= [rows],3= [columns]), also our dimension should be a proper holding same sequence of rows and columns
```

```python
arr2D = np.array([[1, 2, 9], [3, 4, 8], [5, 6, 7], [10, 11, 12], [13, 15]])
print(arr2D, arr2D.shape)


# here [13, 15] is an inappropriate sequence, the list should have same number of values to get a shape
#setting an array element with a sequence. The requested array has an inhomogeneous shape after 1 dimensions.
#The detected shape was (5,) + inhomogeneous part.
```


# Creating arrays using built-in function

## 1) zeros_like
## --> Return an array of zeros with shape and type of input.

```python
# numpy.zeros(shape, dtype=None (by default it prints it as a float), order='C', *, device=None, like=None)
# Return a new array of given shape and type, filled with zeros.

arr1 = np.zeros((2,3), dtype = "int64") # prefill
print(arr1, arr1.shape)

```

## 2) ones
## --> Return a new array setting values to one.


```python

arr2 = np.ones((2,3), dtype = "int64") # prefill
print(arr2, arr2.shape)


# To create a 1D array 
arr3 = np.ones((5,), dtype = "int64") # prefill
print(arr3, arr3.shape)
```

## 3) full
## --> Return a new array of given shape filled with given value.

```python
arr4 = np.full((3,4), 100)  # will fill the given "100" in the given dimention
print(arr4, arr4.shape)
```

## 4) numpy.eye
##  --> Return a 2-D array with ones on the diagonal and zeros elsewhere.

```python
# numpy.eye(N, M=None, k=0, dtype=<class 'float'>, order='C', *, device=None, like=None)
# used for identity matrix = yeh ek square matrix hoti hai, jisme left side diagonal "1" hoti hai baaki poori matrix main "0" fill krti hai 
# --> isme jo value hum pass krte utni ki grid bnti hai (Example- 5 hai toh: 5 rows and 5 column)
arr5 = np.eye(5)
print(arr5, arr5.shape)
```

## 5) numpy.arange
## --> Just like Python's "Range" function  (start, stop, step)

```python
arr6 = np.arange(0, 11, 2)
print(arr6)
```
