## Vectorization

```python
import numpy as np
```

```python
# vectorization - it eliminates the use of a operation on individual input and operates in bulk 

arr1 = np.array([1, 2, 3, 5, 6, 7])
arr2 = np.array([10, 12, 13, 15, 16, 17])

print(arr1 ** 2)
print(arr1 + 10)
print(arr1 + arr2)
```

## Broadcasting

```python
A way to perform operations on arrays of different shapes without manually resizing them.

-----> Instead of copying data, NumPy automatically stretches (broadcasts) the smaller array to match the larger one.


    “If dimensions match or one is 1 → broadcasting works”
```

```python
# Scalar + Array

arr = np.array([1, 2, 3]) # Array
result = arr + 5  # Scalar
print(result)

# Output: [6 7 8]

#  5 is broadcast to [5, 5, 5]    --> cause shape of 5 is "1 x 5" and the array is "1 x 3" ==> both have '1', so will be compatible
```

```python
# Different shapes


arr1 = np.array([[1, 2, 3],[4, 5, 6]])
arr2 = np.array([10, 20, 30])
print(arr1 + arr2)
```

## Normalization

Normalization an array means transforming its value so they fit into a specific scale or range

```python
arr = np.array([[1, 2],[3, 4]])

mean = np.mean(arr)
std_dev = np.std(arr)

normalized_arr = (arr - mean)/std_dev
print(normalized_arr)

print(mean)
print(std_dev)
```

```python
# ab jo humara transformed array hai iska mean '0' aayga and iska standard deviation '1' ke jinte close ho skti hai vo hogi 
```
