```python
import numpy as np
import time #to measure time
```

```python
#to create a numpy array
arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr))
```

```python
# to compare the execution performance of Python vs Numpy
size = 1_000_000

# via python list processing time
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list time = {end - start} seconds")

# via Numpy array processing time
np_arr = np.array(py_list)
start = time.time()

# vectorization - it eliminates the use of a operation on individual input and operates in bulk 
sq_array = np_arr ** 2
end = time.time()
print(f"Numpy array time = {end - start} seconds")
```

```python
# memory 
import sys

print(f"Python list size = {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"Numpy array size = {np_arr.nbytes} bytes")
```
